<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Project в C#, который будет представлять проекты в рамках
организации. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

**Требования к базовому классу Project:**

- **Атрибуты:** ID проекта (ProjectId), Название проекта (ProjectName), Статус
проекта (Status).

- **Методы:**

    - UpdateStatus(): метод для обновления статуса проекта.
    - GetProjectDetails(): метод для получения деталей проекта.
    - AssignTeamMember(): метод для назначения командного члена
проекту.

**Требования к производным классам:**

1. ПродуктовыйПроект (ProductProject): Должен содержать дополнительные
атрибуты, такие как Ожидаемая дата завершения (ExpectedCompletionDate).
Метод UpdateStatus() должен быть переопределен для включения
ожидаемой даты завершения в статус проекта.

2. ИсследовательскийПроект (ResearchProject): Должен содержать
дополнительные атрибуты, такие как Финансирование (FundingAmount).
Метод AssignTeamMember() должен быть переопределен для указания
специфических требований к навыкам командного члена для
исследовательского проекта.

3. ИнфраструктурныйПроект (InfrastructureProject) (если требуется третий
класс): Должен содержать дополнительные атрибуты, такие как Срок
выполнения (ExecutionPeriod). Метод GetProjectDetails() должен быть
переопределен для отображения срока выполнения проекта вместе с
другими деталями проекта.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
using System.Collections.Generic;

public class Project
{
    public int ProjectId { get; private set; }
    public string ProjectName { get; set; }
    public string Status { get; set; }
    public List<string> TeamList { get; set; }
    public string ManagerName { get; set; }
    public DateTime StartDate { get; set; }

    public static int TotalProjects { get; private set; } = 0;

    public Project()
    {
        TeamList = new List<string>();
        TotalProjects++;
        ProjectId = TotalProjects;
        Status = "Планируется";
        StartDate = DateTime.Now;
    }

    public virtual void UpdateStatus(string newStatus)
    {
        Status = newStatus;
        Console.WriteLine($"Статус проекта `{ProjectName}` обновлен на: {Status}");
    }

    public void UpdateStatus(string newStatus, string comment)
    {
        Status = newStatus;
        Console.WriteLine($"Статус `{ProjectName}`: {Status}. Комментарий: {comment}");
    }

    public virtual void GetProjectDetails()
    {
        Console.WriteLine("ID: " + ProjectId);
        Console.WriteLine("Название: " + ProjectName);
        Console.WriteLine("Статус: " + Status);
        Console.WriteLine("Менеджер: " + ManagerName);
        
        if (TeamList.Count == 0)
            Console.WriteLine("Состав команды: Никто не назначен");
        else
            Console.WriteLine("Состав команды: " + String.Join(", ", TeamList));
    }

    public void GetProjectDetails(bool showId)
    {
        if (showId) Console.WriteLine($"ID проекта: {ProjectId}");
        GetProjectDetails();
    }

    public virtual void AssignTeamMember(string memberName)
    {
        TeamList.Add(memberName);
        Console.WriteLine($"Участник {memberName} назначен на проект `{ProjectName}`");
    }

    public static void DisplayTotalProjects()
    {
        Console.WriteLine($"Общее количество проектов: {TotalProjects}");
    }
}

public interface IFunding
{
    void ChangeFunding(int value);
    void PaySalary(int amount);
}

public class ProductProject : Project
{
    private string state;
    public string ExpectedCompletionDate { get; set; }
    public string ProjectResult { get; set; }
    public decimal Budget { get; set; }

    public void SetNewExpectDate(string newDate)
    {
        ExpectedCompletionDate = newDate;
        UpdateStatus(state);
    }

    public override void UpdateStatus(string newStatus)
    {
        state = newStatus;
        Status = state + "; " + ExpectedCompletionDate + " (примерная дата выполнения)";
        base.UpdateStatus(Status);
    }

    public override void GetProjectDetails()
    {
        base.GetProjectDetails();
        Console.WriteLine($"Результат проекта: {ProjectResult}");
    }

    public void OrderResources(int amount)
    {
        Console.WriteLine($"Заказано {amount} единиц ресурсов для проекта {ProjectName}. Бюджет: {Budget}");
    }

    public void OrderResources(string resourceName)
    {
        Console.WriteLine($"Заказан ресурс: {resourceName} для проекта {ProjectName}.");
    }
}

public class ResearchProject : Project, IFunding
{
    public int FundingAmount { get; set; } = 80000;
    public string Requirements { get; set; }
    public string ResearchTheme { get; set; }
    public string ResearchArea { get; set; }

    public void ChangeFunding(int value)
    {
        FundingAmount += value;
        if (value > 0)
            Console.WriteLine($"Бюджет {ProjectName} увеличился на {value} рублей. Теперь: {FundingAmount}");
        else
            Console.WriteLine($"Бюджет {ProjectName} уменьшился на {Math.Abs(value)} рублей. Теперь: {FundingAmount}");
    }

    public void PaySalary(int amount)
    {
        foreach (string person in TeamList)
        {
            if (FundingAmount > amount)
            {
                Console.WriteLine($"{person} получил зарплату в размере {amount} рублей");
                FundingAmount -= amount;
            }
            else
            {
                Console.WriteLine($"{person} получил зарплату в размере {FundingAmount} рублей");
                FundingAmount = 0;
                Console.WriteLine($"У {ProjectName} закончился бюджет на исследования");
                break;
            }
        }
    }

    public override void GetProjectDetails()
    {
        base.GetProjectDetails();
        Console.WriteLine($"Тема исследования: {ResearchTheme}");
        Console.WriteLine($"Бюджет исследования: {FundingAmount} рублей");
    }

    public override void AssignTeamMember(string memberName)
    {
        TeamList.Add(memberName);
        Console.WriteLine($"{memberName}, обладающий навыками: {Requirements}, назначен в проект: {ProjectName}");
    }

    public void AllocateFunding(decimal amount)
    {
        FundingAmount += (int)amount;
        Console.WriteLine($"Выделено {amount} руб. Итого бюджет: {FundingAmount} руб.");
    }

    public void AllocateFunding(decimal amount, string sponsor)
    {
        FundingAmount += (int)amount;
        Console.WriteLine($"Спонсор {sponsor} выделил {amount} руб. Итого: {FundingAmount} руб.");
    }
}

public class InfrastructureProject : Project
{
    public int ExecutionPeriod { get; set; }
    public string Units { get; set; } = "дней";
    public string Location { get; set; }

    public override void GetProjectDetails()
    {
        base.GetProjectDetails();
        Console.WriteLine($"Срок выполнения: {ExecutionPeriod} {Units}. Локация: {Location}");
    }

    public void TakeTeamMembersFromOther(Project otherProject)
    {
        foreach (string teamMember in otherProject.TeamList)
        {
            AssignTeamMember(teamMember);
        }
    }

    public void CheckSafety(bool passed)
    {
        Console.WriteLine(passed ? "Проверка безопасности пройдена." : "ВНИМАНИЕ: Проверка безопасности не пройдена!");
    }
}

public class ProgramProject : ProductProject
{
    public string Language { get; set; }

    public override void GetProjectDetails()
    {
        base.GetProjectDetails();
        Console.WriteLine($"Язык программирования: {Language}");
    }
}

public class ProjectCollection<T> where T : Project
{
    private List<T> _projects = new List<T>();

    public void Add(T project) => _projects.Add(project);
    public void Remove(T project) => _projects.Remove(project);

    public void DisplayProjects()
    {
        Console.WriteLine("Список проектов в коллекции: ");
        foreach (var project in _projects)
            Console.WriteLine($"- {project.ProjectName} (ID: {project.ProjectId})");
    }
}

var productProject = new ProductProject
{
    ProjectName = "Мобильное приложение",
    ExpectedCompletionDate = "15 ноября",
    ProjectResult = "Приложение для доставки",
    ManagerName = "Иванов И.И.",
    Budget = 100000
};
productProject.AssignTeamMember("Смирнов А.");
productProject.UpdateStatus("В разработке");
productProject.OrderResources(50);
productProject.OrderResources("Серверные мощности");

var researchProject = new ResearchProject
{
    ProjectName = "Квантовые вычисления",
    FundingAmount = 120000,
    Requirements = "Физика, Математика",
    ResearchTheme = "Квантовые алгоритмы",
    ManagerName = "Петров П.П.",
    ResearchArea = "Квантовая физика"
};
researchProject.AssignTeamMember("Козлов Д.");
researchProject.ChangeFunding(30000);
researchProject.PaySalary(7000);
researchProject.AllocateFunding(50000, "Государственный фонд");

var infraProject = new InfrastructureProject
{
    ProjectName = "Ремонт офиса",
    ExecutionPeriod = 2,
    Units = "недели",
    ManagerName = "Сидоров С.С.",
    Location = "Москва, ул. Ленина"
};
infraProject.AssignTeamMember("Петров В.");
infraProject.TakeTeamMembersFromOther(researchProject);
infraProject.CheckSafety(true);

productProject.GetProjectDetails(true);
researchProject.GetProjectDetails();
infraProject.GetProjectDetails();

var collection = new ProjectCollection<Project>();
collection.Add(productProject);
collection.Add(researchProject);
collection.Add(infraProject);

Console.WriteLine();
collection.DisplayProjects();

Console.WriteLine();
Project.DisplayTotalProjects();
```
